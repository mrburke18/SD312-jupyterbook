```python
import numpy as np

N=25

data=np.random.rand(N,2)*2-1
label=np.zeros(N)
for i in range(N):
    label[i]=np.linalg.norm(data[i,:])<.75
```

```python
import plotly.graph_objects as go
fig=go.Figure(data=go.Scatter(x=data[:,0],y=data[:,1],mode='markers',
                              marker=dict(color=label,colorscale='Portland')))
fig.show()
```

```python
np.savez('nonseparable.npz',array1=data,array2=label)
```
