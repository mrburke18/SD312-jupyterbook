```python
# Define the S3 bucket name and the key (path) of the tar.gz file
bucket_name = 'mlspace-data-521454461163'
dataset_path = 'private/taylor/datasets/fishparakeet' #TODO HERE
key = 'fishparakeet.tgz' #TODO HERE
```

```python
# This example code downloads a tarfile from S3. The code then uploads the extracted files to S3.

import boto3
import tarfile
import os

# Initialize a boto3 S3 client
s3 = boto3.client('s3')

# Define the local path to save the downloaded file
local_filename = '/home/ec2-user/SageMaker/tarfolder.tar.gz'


# Download the tar.gz file from S3
s3.download_file(bucket_name, f'{dataset_path}/{key}', local_filename)

# Extract the downloaded tar.gz file to the local folder
with tarfile.open(local_filename, "r:gz") as tar_ref:
    tar_ref.extractall('/home/ec2-user/SageMaker/imgs')

# Optionally, remove the downloaded tar.gz file if not needed
os.remove(local_filename)

# Re-upload the decompressed files to S3

client = boto3.client('s3')
local_directory = '/home/ec2-user/SageMaker/imgs'
bucket_name = 'mlspace-data-521454461163'

#Optionally add a folder name here for S3. This will be the prefix folder in S3
destination = dataset_path

for root, dirs, files in os.walk(local_directory):
    for filename in files:
        # Construct the full local path
        local_path = os.path.join(root, filename)

        # Construct the full S3 path
        relative_path = os.path.relpath(local_path, local_directory)
        s3_path = os.path.join(destination, relative_path)

        # Upload the file
        print(f"Uploading {s3_path} to {bucket_name}...")
        client.upload_file(local_path, bucket_name, s3_path)
```
