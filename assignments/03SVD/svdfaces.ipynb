# Compression via the SVD and low-rank approximation

As a reminder and reference, we can decompose any matrix into its SVD:

$$\underbrace{A}_{n \times m} = \underbrace{U}_{n \times n}~ \underbrace{\Sigma}_{n \times m}~ \underbrace{V^T}_{m \times m}.$$

$\Sigma$ is a diagonal matrix, with decreasing values, indicating that the progressive columns of $U$ and rows of $V^T$ become less and less significant in reconstructing the data. This allows us to create a so-called "low-rank approximation," where we take our matrix $A$, and approximate it with a matrix $\hat{A}$, which is of much lower rank than $A$ (call it rank $k$). We can do this by zeroing out the smallest values of $\Sigma$, thereby removing the least significant columns of $U$ and rows of $V^T$ from the calculation (because they're just multiplied by 0 anyway).

$$\underbrace{\hat A}_{n \times m} = \underbrace{\hat U}_{n \times k} ~\underbrace{\hat \Sigma}_{k \times k} ~\underbrace{\hat V^T}_{k \times m}.$$

One natural question is, how much are we losing when we perform this low-rank approximation?
