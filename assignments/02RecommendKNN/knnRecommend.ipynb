We're going to create a movie recommendation system that will recommend a movie similar to an input movie.  In other words, perhaps a person just finished watching the horror movie "Saw."  We're going to assume they liked it, and recommend another movie like "Saw" by using only user ratings (we'll consider the individual user's actual preferences and previous movie ratings in a later project).

You will need both `ratings.csv` and `movies.csv` to do this.
